```python
from transformers import Qwen2_5_VLForConditionalGeneration, AutoTokenizer, AutoProcessor
from qwen_vl_utils import process_vision_info
import torch

qwen_32b = "Qwen/Qwen2.5-VL-32B-Instruct"

# default: Load the model on the available device(s)
model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    qwen_32b, torch_dtype="auto", device_map="auto"
)

# We recommend enabling flash_attention_2 for better acceleration and memory saving, especially in multi-image and video scenarios.
# model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
#     qwen_32b,
#     torch_dtype=torch.bfloat16,
#     attn_implementation="flash_attention_2",
#     device_map="auto",
# )

# default processer
processor = AutoProcessor.from_pretrained("Qwen/Qwen2.5-VL-32B-Instruct")

# The default range for the number of visual tokens per image in the model is 4-16384.
# You can set min_pixels and max_pixels according to your needs, such as a token range of 256-1280, to balance performance and cost.
# min_pixels = 256*28*28
# max_pixels = 1280*28*28
# processor = AutoProcessor.from_pretrained("Qwen/Qwen2.5-VL-32B-Instruct", min_pixels=min_pixels, max_pixels=max_pixels)


USER_PROMPT = """
Extract the Jawi in the file with the following format
<jawi text>
<Malay romanization>
"""

messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "image",
                # "image": "https://qianwen-res.oss-cn-beijing.aliyuncs.com/Qwen-VL/assets/demo.jpeg",
                "image": project_path / "sample_jawi.PNG"
            },
            {"type": "text", "text": USER_PROPMT},
        ],
    }
]

# Preparation for inference
text = processor.apply_chat_template(
    messages, tokenize=False, add_generation_prompt=True
)
image_inputs, video_inputs = process_vision_info(messages)
inputs = processor(
    text=[text],
    images=image_inputs,
    videos=video_inputs,
    padding=True,
    return_tensors="pt",
)
inputs = inputs.to("cuda")

# Inference: Generation of the output
generated_ids = model.generate(**inputs, max_new_tokens=128)
generated_ids_trimmed = [
    out_ids[len(in_ids) :] for in_ids, out_ids in zip(inputs.input_ids, generated_ids)
]
output_text = processor.batch_decode(
    generated_ids_trimmed, skip_special_tokens=True, clean_up_tokenization_spaces=False
)
print(output_text)
```

```python
from PIL import Image, ImageDraw, ImageFont
from pathlib import Path
from tqdm.notebook import tqdm
import requests
from bs4 import BeautifulSoup
from requests.structures import CaseInsensitiveDict


def generate_khat(jawi_story_sentences:list, fontfile):
    # Specify the path to your TrueType font file (.ttf or .otf)
    # font_path = "/content/drive/MyDrive/Colab Notebooks/MALAYSIA - AI/A Thuluth Regular.ttf"  # Replace with the actual path
    font_path = f"/content/drive/MyDrive/Colab Notebooks/MALAYSIA - AI/{fontfile}"  # Replace with the actual path
    if not Path(font_path).exists():
        # print("font path does not exist!")
        raise Exception("Font does not exist!")
    else:
        pass

    font_size = 30
    text_color = (0, 0, 0)  # Black
    background_color = (255, 255, 255)  # White
    padding = 40 # Add some padding around the text

    # Load font once
    try:
        font = ImageFont.truetype(font_path, font_size)
    except IOError:
        print(f"Could not load font: {font_path}. Using default font.")
        font = ImageFont.load_default()

    for i in tqdm(range(len(jawi_story_sentences))):
        current_text = jawi_story_sentences[i]

        # Temporarily create a dummy ImageDraw object to calculate text size
        # This is a common workaround when you need textbbox without an actual image yet
        dummy_img = Image.new('RGB', (1, 1))
        dummy_draw = ImageDraw.Draw(dummy_img)

        # Get text bounding box
        # The bbox method provides (left, top, right, bottom) coordinates relative to the text's origin.
        # Note: getbbox might return (x0, y0, x1, y1) where x0/y0 can be negative for characters
        # that extend left/above the origin. We need the total width/height.
        text_bbox = dummy_draw.textbbox((0, 0), current_text, font=font)

        # Calculate actual text width and height from the bounding box
        text_width = text_bbox[2] - text_bbox[0]
        text_height = text_bbox[3] - text_bbox[1]

        # Calculate image dimensions with padding
        image_width = text_width + (2 * padding)
        image_height = text_height + (2 * padding)

        # Create a new blank image with dynamic dimensions
        img = Image.new('RGB', (image_width, image_height), color=background_color)
        draw = ImageDraw.Draw(img)

        # Calculate coordinates to center the text with padding
        # We use -text_bbox[0] and -text_bbox[1] to adjust for potential negative offsets
        # from textbbox, ensuring the text starts correctly from the padding.
        x = padding - text_bbox[0]
        y = padding - text_bbox[1]

        draw.text((x, y), current_text, font=font, fill=text_color)

        # Draw bounding boxes for each character
        # TODO (problem)
        box_color = (255, 0, 0)  # Red for bounding boxes
        char_x = x
        for char in current_text:
            char_bbox = draw.textbbox((char_x, y), char, font=font)
            draw.rectangle(char_bbox, outline=box_color, width=1)
            char_width = char_bbox[2] - char_bbox[0]
            char_x += char_width  # Move to next character position

        output_filename = f"output_image_{i}.png"
        img.save(out_path / output_filename)

def get_jawi(string):
    url = "https://www.ejawi.net/result.php"
    headers = CaseInsensitiveDict()
    headers["authority"] = "www.ejawi.net"
    headers["accept"] = "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8,application/signed-exchange;v=b3;q=0.9"
    headers["accept-language"] = "en-MY,en;q=0.9,en-US;q=0.8,ms;q=0.7"
    headers["cache-control"] = "no-cache"
    headers["content-type"] = "application/x-www-form-urlencoded"
    headers["cookie"] = "__utma=248434904.1854521706.1653239537.1653239537.1653239537.1; __utmc=248434904; __utmz=248434904.1653239537.1.1.utmcsr=google|utmccn=(organic)|utmcmd=organic|utmctr=(not%20provided); __utmt=1; __utmb=248434904.1.10.1653239537"
    headers["origin"] = "https://www.ejawi.net"
    headers["pragma"] = "no-cache"
    headers["referer"] = "https://www.ejawi.net/converterV2.php?go=rumi"
    headers["sec-ch-ua"] = '" Not A;Brand";v="99", "Chromium";v="100", "Google Chrome";v="100"'
    headers["sec-ch-ua-mobile"] = "?0"
    headers["sec-ch-ua-platform"] = '"macOS"'
    headers["sec-fetch-dest"] = "iframe"
    headers["sec-fetch-mode"] = "navigate"
    headers["sec-fetch-site"] = "same-origin"
    headers["sec-fetch-user"] = "?1"
    headers["upgrade-insecure-requests"] = "1"
    headers["user-agent"] = "Mozilla/5.0 (Macintosh; Intel Mac OS X 10_15_2) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/100.0.4896.88 Safari/537.36"
    data = f"phonText={string}&disable_direction=on&number=1"
    resp = requests.post(url, headers=headers, data=data)
    decoded = resp._content.decode()
    soup = BeautifulSoup(decoded, 'html.parser')
    return soup.find_all('td')[0].text

def clean(text:str):
    return text.replace("'", '').replace('﴿','(').replace('﴾',')')
```

```python
from PIL import Image, ImageDraw
import freetype
import uharfbuzz as hb
import arabic_reshaper
from bidi.algorithm import get_display
from pathlib import Path
from tqdm import tqdm

def generate_khat_v2(jawi_story_sentences, fontfile, out_path=Path("out_path"), font_size=60):
    font_path = f"/content/drive/MyDrive/Colab Notebooks/MALAYSIA - AI/{fontfile}"
    if not Path(font_path).exists():
        raise Exception("Font does not exist!")

    face = freetype.Face(font_path)
    face.set_char_size(font_size * 64)

    with open(font_path, "rb") as font_file:
        font_data = font_file.read()

    hb_font = hb.Font(hb.Face(font_data))
    scale = face.size.height >> 6

    for idx, sentence in enumerate(tqdm(jawi_story_sentences)):
        reshaped = arabic_reshaper.reshape(sentence)
        bidi_text = get_display(reshaped)

        buf = hb.Buffer()
        buf.add_str(bidi_text)
        buf.guess_segment_properties()
        hb.shape(hb_font, buf)

        infos = buf.glyph_infos
        positions = buf.glyph_positions

        width = sum([pos.x_advance for pos in positions]) // 64
        height = scale

        padding = 50
        img = Image.new("RGB", (width + 2 * padding, height + 2 * padding), (255, 255, 255))
        draw = ImageDraw.Draw(img)

        x, y = padding, padding
        for info, pos in zip(infos, positions):
            gid = info.codepoint
            face.load_glyph(gid, freetype.FT_LOAD_RENDER)
            bitmap = face.glyph.bitmap
            top = face.glyph.bitmap_top
            left = face.glyph.bitmap_left

            glyph_width = bitmap.width
            glyph_height = bitmap.rows

            # Paste the glyph bitmap into the image
            glyph_img = Image.frombytes("L", (glyph_width, glyph_height), bitmap.buffer)
            img.paste(Image.merge("RGB", (glyph_img, glyph_img, glyph_img)), (x + left, y + (scale - top)))

            # Draw bounding box
            draw.rectangle([x + left, y + (scale - top),
                            x + left + glyph_width, y + (scale - top) + glyph_height],
                           outline=(255, 0, 0), width=1)

            x += pos.x_advance // 64

        img.save(out_path / f"output_image_{idx}.png")
```
